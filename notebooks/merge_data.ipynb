```python
import pandas as pd

# 1. Load fitur dari dataset
df_features = pd.read_excel(r'C:\Folder Dzakwan\Folder Dzakwan\Keperluan TA\dataset\primary_dataset_features.xlsx')

# 2. Load hasil - PENTING: Tambahkan parameter sheet_name='HSV Masking'
df_hasil = pd.read_excel(r'C:\Folder Dzakwan\Folder Dzakwan\Keperluan TA\dataset\Hasil Pengambilan Data.xlsx', sheet_name='HSV Masking')

# 3. Ambil angka ID dari kolom 'Subjek' (Misal: 'Subj_1' -> 1)
df_features['Subj_ID'] = df_features['Subjek'].str.extract(r'(\d+)').astype(int)

# 4. Ambil kolom yang dibutuhkan dari hasil dan bersihkan baris kosong
df_hasil_subset = df_hasil[['Subj_ID', 'Jenis Kelamin', 'Hasil Pengukuran']].dropna(subset=['Subj_ID'])
df_hasil_subset['Subj_ID'] = df_hasil_subset['Subj_ID'].astype(int)

# 5. Lakukan proses merge/mapping berdasarkan 'Subj_ID'
df_merged = pd.merge(df_features, df_hasil_subset, on='Subj_ID', how='left')

# 6. Timpa kolom 'Gender' dan 'Hemoglobin' yang kosong dengan data yang baru
df_merged['Gender'] = df_merged['Jenis Kelamin']
df_merged['Hemoglobin'] = df_merged['Hasil Pengukuran']

# 7. Hapus kolom bantuan yang sudah tidak terpakai
df_merged = df_merged.drop(columns=['Subj_ID', 'Jenis Kelamin', 'Hasil Pengukuran'])

# 8. Cek apakah masih ada yang kosong
print("Jumlah Gender kosong:", df_merged['Gender'].isna().sum())
print("Jumlah Hemoglobin kosong:", df_merged['Hemoglobin'].isna().sum())

# 9. Simpan hasilnya ke file baru
df_merged.to_excel(r'C:\Folder Dzakwan\Folder Dzakwan\Keperluan TA\dataset\primary_dataset_features_lengkap.xlsx', index=False)
print("Selesai! Data berhasil diperbarui.")
```
